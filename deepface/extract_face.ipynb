```python
import os
import glob
import cv2
from deepface import DeepFace
from tqdm import tqdm
```

```python
import tensorflow as tf

# TF 2.1 이상 권장 방식
gpus = tf.config.list_physical_devices('GPU')
print("GPU 장치 목록:", gpus)
print("GPU 사용 가능 여부:", len(gpus) > 0)
```

```python
input_dir = '/home/ckim/extracted_frames'
output_dir = '/home/ckim/cropped_faces'
os.makedirs(output_dir, exist_ok=True)
```

```python
# 프레임 이미지 리스트 가져오기
frame_files = sorted(glob.glob(os.path.join(input_dir, '*.jpg')))
```

```python
for img_path in tqdm(frame_files, desc="정상 얼굴만 저장 중"):
    filename = os.path.basename(img_path)
    img = cv2.imread(img_path)

    if img is None:
        print(f"{filename} 로드 실패")
        continue

    try:
        img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        result = DeepFace.analyze(img_rgb, detector_backend='retinaface', enforce_detection=False)

        result = result[0] if isinstance(result, list) else result
        region = result['region']
        x, y, w, h = region['x'], region['y'], region['w'], region['h']

        # 유효성 필터링: 너무 크거나 너무 작은 얼굴은 제외
        img_h, img_w = img.shape[:2]
        face_area = w * h
        img_area = img_w * img_h

        if face_area < 100*100 or face_area > 0.8 * img_area:
            print(f"{filename}: 얼굴 크기 비정상 (스킵)")
            continue

        face_crop = img[y:y+h, x:x+w]
        cv2.imwrite(os.path.join(output_dir, filename), face_crop)

    except Exception as e:
        print(f"{filename} 처리 중 오류: {e}")
```
